```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
from min_tree import weight_complete_graph, minimum_spanning_tree
from nqueens import n_queens_tree
```

```python
g = nx.Graph()
g.add_nodes_from(["a", "b", "c"])
g.add_weighted_edges_from([("a", "b", 2), ("a", "c", 0.5)])
position=nx.spring_layout(g)
weight_labels = nx.get_edge_attributes(g, 'weight')
nx.draw_networkx_edge_labels(g, pos=position,
                             edge_labels=weight_labels,
                             font_color='red')
nx.draw(g, with_labels="True", pos=position)
```

```python
#Generamos una grafica completa de n nodos y pesos aleatorios
n = 7
G = weight_complete_graph(n)
position=nx.spring_layout(G)
weight_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos=position,
                             edge_labels=weight_labels,
                             font_color='red')
nx.draw(G, with_labels="True", pos=position)
```

```python
#Calcula el arbol generador a partir de la funcion de networkx
mtgnx= nx.minimum_spanning_tree(G)
mtg_position=nx.spring_layout(mtgnx)
weight_labels = nx.get_edge_attributes(mtgnx, 'weight')
nx.draw_networkx_edge_labels(mtgnx, pos=mtg_position, edge_labels=weight_labels, font_color='red')
nx.draw(mtgnx, with_labels="True", pos=mtg_position)
```

```python
#Encuentra el arbol generador minimo a parti de nuestra funcion
mtg = minimum_spanning_tree(G)
mtg_position=nx.spring_layout(mtg)
weight_labels = nx.get_edge_attributes(mtg, 'weight')
nx.draw_networkx_edge_labels(mtg, pos=mtg_position, edge_labels=weight_labels, font_color='red')
nx.draw(mtg, with_labels="True", pos=mtg_position)
```
