```python
class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))
        self.rank = [0] * size

    def find(self, p):
        if self.parent[p] != p:
            self.parent[p] = self.find(self.parent[p])
        return self.parent[p]

    def union(self, p, q):
        rootP = self.find(p)
        rootQ = self.find(q)
        if rootP == rootQ:
            return False
        if self.rank[rootP] < self.rank[rootQ]:
            self.parent[rootP] = rootQ
        elif self.rank[rootP] > self.rank[rootQ]:
            self.parent[rootQ] = rootP
        else:
            self.parent[rootQ] = rootP
            self.rank[rootP] += 1
        return True

def kruskal_mst(G):
    """
    Encuentra el árbol generador de menor costo usando NetworkX.
    G: grafo NetworkX con pesos en las aristas.
    Retorna el MST como un grafo NetworkX.
    """
    return nx.minimum_spanning_tree(G)

# Ejemplo de uso
if __name__ == "__main__":
    n = 5
    G = generate_complete_graph(n)
    print("Grafo generado:", G.edges(data=True))
    mst = kruskal_mst(G)
    print("MST:", mst.edges(data=True))
```
